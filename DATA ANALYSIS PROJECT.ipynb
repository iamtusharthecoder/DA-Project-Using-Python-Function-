```python
# VISUALIZATION
```

```python
# hostogram

sns.histplot(data['PRICE PER BOX'])
plt.grid(True)
plt.show()
 
```

```python
# Heatmap

corr = data.select_dtypes(include='number').corr()
sns.heatmap(corr,annot=True)
plt.show()
```

```python
# country-wise price

sns.barplot(x='COUNTRY', y='PRICE PER BOX',data=data)
plt.xticks(rotation=90)
plt.grid(True)
plt.show()
```

```python
# kdeplot

sns.kdeplot(data=data, x="MARKETING SPEND",fill=True)
plt.grid(True)
plt.show()
```

```python
# KEY ANALYTICAL QUESTION AND INSIGHTS
```

```python
# 1 Which are the Top 5 countries by sales?

top_countries = data.groupby("COUNTRY")["AMOUNT"].sum().sort_values(ascending=False).head(10)

print(top_countries)
```

```python
#2 Which sales channel generated the highest revenue?

channel_sales = data.groupby("CHANNEL")["AMOUNT"].sum().sort_values(ascending=False)

print(channel_sales)
```

```python
# 3 Which products shipped the most boxes?

top_boxes = data.groupby("PRODUCT")["BOXES SHIPPED"].sum().sort_values(ascending=False).head(10)

print(top_boxes)
```

```python
# 4  Which country placed the highest number of orders?

country_orders = data.groupby("COUNTRY")["ORDER ID"].count().sort_values(ascending=False)

print(country_orders)
```
